This notebook runs the reproducible forecasting pipeline in `src/forecast_adamw.py`, reviews its holdout performance, and inspects the September 2026–February 2027 projections.

**Important:** AdamW is the optimizer, not the forecasting algorithm. The forecasting algorithm here is a compact PyTorch feed-forward neural network trained with AdamW.

```python
from pathlib import Path
import runpy
ROOT = Path.cwd()
if not (ROOT / 'src/forecast_adamw.py').exists():
    ROOT = Path('../').resolve()
assert (ROOT / 'src/forecast_adamw.py').exists(), 'Run this notebook from the project root or notebooks folder.'
runpy.run_path(str(ROOT / 'src/forecast_adamw.py'), run_name='__main__')
```

## Inspect the forecast and validation
The holdout is the final 28 observed days. Compare AdamW MAE with the seven-day-mean naive benchmark before interpreting the six-month output.

```python
import json
import pandas as pd
meta = json.loads((ROOT / 'outputs/adamw_forecast_metadata.json').read_text())
forecast = pd.read_csv(ROOT / 'outputs/adamw_six_month_forecast.csv')
display(pd.DataFrame([{'Metric':'AdamW holdout MAE (₹/day)','Value':meta['holdout_mae_rupees_per_day']}, {'Metric':'Naive 7-day mean holdout MAE (₹/day)','Value':meta['naive_7day_mean_holdout_mae_rupees_per_day']}, {'Metric':'Holdout WMAPE (%)','Value':meta['holdout_wmape_pct']}, {'Metric':'AdamW beats naive?','Value':meta['beats_naive_baseline_on_holdout_mae']}]))
display(forecast)
```

## Interpretation
If AdamW does not beat the naive baseline, do not present it as the preferred production model. The forecast is gross recorded recovery, not incremental recovery, net recovery, or ROI. The current history is too short to learn annual seasonality, and the indicative error bands are not calibrated prediction intervals.
